```python
# Mi meta de ahorro - quiero juntar para un viaje
ahorro = 500000       # <- lo que ya tienes
meta = 3000000      # <- cambia, mi meta es 5000
por_mes = 80000    # <- cambia, lo que ahorro al mes
meses = 0

# Mientras no llegues a la meta:
# suma lo del mes y cuenta un mes más
while ahorro < meta:
    ahorro = ahorro + por_mes
    meses = meses + 1
    print(ahorro)

# ¿Cuántos meses tardaste?
print("Meses:", meses)
```

```python
# Dividir la cuenta - salimos a comer
cuenta = 350000   # <- cambia, lo que gastamos
propina = 6    # <- cambia (%), lo que dejamos de propina
amigos = 5       # <- cambia, cuántos somos

# 1. Regresa la cuenta más la propina
def con_propina(cuenta, porcentaje):
    return cuenta + cuenta * porcentaje / 100

# 2. Regresa cuánto paga cada persona
def por_persona(total, personas):
    return total / personas

# 3. Úsalas con tus datos e imprime
total_con_propina = con_propina(cuenta, propina)
print("Total con propina:", total_con_propina)

cada_uno = por_persona(total_con_propina, amigos)
print("Cada uno paga:", cada_uno)

# Función nueva
def descuento(total, porcentaje):
    return total - total * porcentaje / 100

print("Con descuento del 10%:", descuento(cuenta, 10))
```

```python
# La caja de la tienda - mi compra
precios = [12000, 4500, 30000, 8000]  # <- mis 4 cosas: chocolate, pan, café, leche

# 1. Suma el total con un for
total = 0
for precio in precios:
    total = total + precio

# 2. Si el total pasa de 500, aplica 10 % de descuento
descuento = 0
if total > 500:
    descuento = total * 10 / 100

a_pagar = total - descuento

# 3. Imprime el ticket con f-strings
print(f"Total: {total}")
print(f"Descuento: {descuento}")
print(f"A pagar: {a_pagar}")
```
