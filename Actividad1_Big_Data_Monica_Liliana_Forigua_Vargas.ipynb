```python
import pandas as pd

# Creamos la tabla de ventas
ventas = pd.DataFrame({
    'fecha': ['2026-11-20', '2026-11-21', '2026-11-22', '2026-11-23', '2026-11-24', '2026-11-25'],
    'canal': ['Tienda', 'Web', 'App', 'Tienda', 'Web', 'App'],
    'ciudad': ['Bogotá', 'Bogotá', 'Medellín', 'Medellín', 'Bogotá', 'Cali'],
    'total': [120000, 80000, 150000, 90000, 110000, 70000]
})
```

```python
import pandas as pd

# 1. ¿Cuál es el total vendido por canal? (Tienda, Web, App)
print(ventas.groupby('canal')['total'].sum())

# 2. ¿Cuál es el ticket promedio por ciudad?
print(ventas.groupby('ciudad')['total'].mean())

# 3. ¿Qué día de la semana se vende más?
ventas['fecha'] = pd.to_datetime(ventas['fecha'])
ventas['dia'] = ventas['fecha'].dt.day_name()
print(ventas.groupby('dia')['total'].sum())
print("El día que más se vende es:", ventas.groupby('dia')['total'].sum().idxmax())
```
