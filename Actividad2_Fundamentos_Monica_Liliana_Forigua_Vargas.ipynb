```python
# Función que clasifica la nota
def clasificar_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota < 4.5:
        return "Aprueba"
    else:
        return "Excelente"

# Lista de notas
notas = [2.5, 3.0, 3.8, 4.7, 5.2, 3.0]

# Aplicarla con un for
aprobados = 0
for nota in notas:
    resultado = clasificar_nota(nota)
    print(f"Nota {nota}: {resultado}")
    if resultado == "Aprueba" or resultado == "Excelente":
        aprobados = aprobados + 1

print(f"Estudiantes que aprobaron: {aprobados}")
```

```python
  # Carrito: producto -> (precio, cantidad)
carrito = {
    'camiseta': (45000, 2),
    'jean': (120000, 1),  'medias': (12000, 3)
}

# Recorrer el diccionario y calcular el subtotal
subtotal = 0
for producto, (precio, cantidad) in carrito.items():
    total_producto = precio * cantidad
    print(f"{producto}: ${precio} x {cantidad} = ${total_producto}")
    subtotal = subtotal + total_producto

print(f"Subtotal: ${subtotal}")

# Si supera 200.000, aplica 10% de descuento
descuento = 0
if subtotal > 200000:
    descuento = subtotal * 0.10
    print(f"Descuento 10%: -${descuento}")

subtotal_con_descuento = subtotal - descuento

# Suma IVA del 19%
iva = subtotal_con_descuento * 0.19
total = subtotal_con_descuento + iva

print(f"IVA 19%: ${iva}")
print(f"Total a pagar: ${total:,.0f}")
```

```python
import pandas as pd

# Tabla de empleados que nos dieron
empleados = pd.DataFrame({
    'nombre': ['Milena', 'Carlos', 'Lorena'],
    'area': ['Ventas', 'Ventas', 'Operaciones'],
    'salario': [2500000, 3000000, 2000000]
})

# 1. Agrega 3 empleados nuevos usando pd.concat
nuevos = pd.DataFrame({
    'nombre': ['Diego', 'Sofia', 'Juan'],
    'area': ['Ventas', 'Operaciones', 'Ventas'],
    'salario': [2800000, 2200000, 3200000]
})

empleados = pd.concat([empleados, nuevos], ignore_index=True)
print(empleados)

# 2. Crea la columna bono: 10% si es Ventas, 5% si no
def calcular_bono(fila):
    if fila['area'] == 'Ventas':
        return fila['salario'] * 0.10
    else:
        return fila['salario'] * 0.05

empleados['bono'] = empleados.apply(calcular_bono, axis=1)
print(empleados)

# 3. Calcula el costo total de nómina mensual por área (salario + bono)
empleados['total'] = empleados['salario'] + empleados['bono']
print(empleados.groupby('area')['total'].sum())

# 4. ¿Quién es el empleado con mayor salario?
mayor = empleados.loc[empleados['salario'].idxmax()]
print(f"El de mayor salario es: {mayor['nombre']} con ${mayor['salario']}")
```
